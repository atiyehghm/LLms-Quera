## Batch and Stream

#### Batch

```python
system = SystemMessage(content="تو یک معلم زبان اسپانیایی هستی و هر لغت یا مساله ای که ازت پرسیده میشه رو با مثال توضیح میدی.")

messages = [
    [system, HumanMessage(content="کلمه‌ی پدر به اسپانیایی چی می‌شه؟")],
    [system, HumanMessage(content="چطوری به اسپانیایی سلام و احوالپرسی کنم؟")]
]

response = chat.batch(messages)
```

```python
for res in response:
    print(res.content)
    print('*'*70)
```

```python
import markdown
from IPython.display import display, HTML

def print_md(content):
    content = markdown.markdown(content)
    content = "<div dir=rtl>{}</div>".format(content)
    display(HTML(content))
```

```python
for res in response:
    print_md(res.content)
    print('*'*70)
```

#### Stream

```python
system_ = SystemMessage(content="تو یک معلم زبان اسپانیایی هستی و هر لغت یا مساله ای که ازت پرسیده میشه رو با مثال توضیح میدی.")

messages = [system_, HumanMessage(content="کلمه‌ی پدر به اسپانیایی چی می‌شه؟")]

for chunk in chat.stream(messages):
    print(chunk.content, end="", flush=True)
```

## Prompt Templates

#### prompt template in llms

```python
import getpass
import os
from langchain_cohere.llms import Cohere
from langchain_core.prompts import PromptTemplate
```

```python
from langchain.llms import Cohere
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

model = Cohere(cohere_api_key=API_KEY)
prompt_template = PromptTemplate.from_template(
    "Generate a very short conversation between two people about {topic}"
)
chain = LLMChain(llm=model, prompt=prompt_template)
response = chain.run(topic="weather")
print(response)
```

#### prompt templates in chat

```python
from langchain_cohere import ChatCohere
from langchain_core.prompts import ChatPromptTemplate

chat_model = ChatCohere(cohere_api_key=API_KEY)
```

```python
message_prompt = ChatPromptTemplate.from_messages([
    ("system", "تو یک معلم ریاضی هستی"),
    ("human", "{first_number} به علاوه‌ی {second_number} چند می‌شود؟")
])
```

```python
chat_ai = chat_model.invoke(message_prompt.invoke({"first_number": "دویست و پنجاه", "second_number": "دویست و هفتاد و شش"}))
print(chat_ai.content)
```
